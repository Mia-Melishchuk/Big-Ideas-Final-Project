# Parking 
For my dataset, I decided to look at the true determinant of safety... whether or not you are safe from the Pittsburgh Parking Authority. The selected dataset of parking meters contains data about them in Pittsburgh. Each row in the dataset includes the area, hours, parking length and rates. We will analyze this dataset to see which neighborhoods have the best bang for your buck when it comes to parking.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data = pd.read_csv("Parkingmeters.csv")

def convert_parkinglen(x):
    if "unlimited" in x.lower():
        return 24
    else:
        x_lower = str(x).lower()
        return float(x_lower.split()[0])

data['parkinglen_numeric'] = data['parkinglen'].apply(convert_parkinglen)
grouped = data.groupby('area')['parkinglen_numeric'].first()
grouped.plot.bar(title="Total Parking Length by Area", color='green', figsize=(8,5))
plt.xlabel("Area")
plt.yticks(np.arange(0, 25, 2)) 
plt.ylabel("Total Parking Length (hours)")
plt.show()
```

# Parking Length
The following bar graph organizes each area in the dataset by total parking length. Most values fell into three categories: 2 hour parking, 10 hour parking and 24 hour parking (unlimited parking). With this data, we can find out which areas have the best parking length, however this graph can be hard to read even with the included yticks code to address the y axis.  Thus, we organize it into a dictionary.

```python
import pandas as pd

data = pd.read_csv("Parkingmeters.csv")
dic = {}
def convert_parkinglen(x):
    if "unlimited" in x.lower():
        strx = '24 Hours'
        return strx
    else:
        x_str = x
        return x_str
for index, row in data.iterrows():
    area = row['area']
    hours = convert_parkinglen(row['parkinglen'])
    if area not in dic:
        dic[area] = []
        dic[area].append(hours)

for area, hours_list in dic.items():
    print(area + ":", *hours_list)
```

To see our data better, we organized the dataset through a dictionary to assign each area to its hour value before printing it out. In this way, each area had its value neatly assigned to it and printed out for data analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
data = pd.read_csv("Parkingmeters.csv")

def convert_parkinglen(x):
    if "dynamic" in x.lower():
        return 0
    else:
        x_str = str(x).replace('$', '')
        return float(x_str)
        
data['parkinglen_numeric'] = data['rates'].apply(convert_parkinglen)
grouped = data.groupby('area')['parkinglen_numeric'].first()
grouped.plot.bar(title="Rates", color='green', figsize=(8,5))
plt.xlabel("Area")
plt.yticks(np.arange(0, 10, .5)) 
plt.ylabel("Rates (in dollars)")
plt.show()
```

Looking at our data, we want to isolate the most optimal values in order to determine what area is the "best". To do this, we looked at the areas that allowed for unlimited parking and filtered only for those values.

```python
import pandas as pd

data = pd.read_csv("Parkingmeters.csv")
dic = {}
def convert_parkinglen(x):
    if "unlimited" in x.lower():
        strx = '24 Hours'
        return strx
    else:
        x_str = x
        return x_str
for index, row in data.iterrows():
    area = row['area']
    hours = convert_parkinglen(row['rates'])
    if area not in dic:
        dic[area] = []
        dic[area].append(hours)
        
for area, hours_list in dic.items():
    curr = hours_list[0]
    if(curr == '$1.00'):
        print(area + ":", *hours_list)
```

In a similar manner, we applied it to rates. Although rates at 1.00 dollar exist, 2 dollars ended up providing the most optimal rate when comparing areas with unlimited parking lengths. Thus, it was found that the best neighborhood was Bakery Square!
